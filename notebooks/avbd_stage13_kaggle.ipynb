# AVBD Stage 13 — allocation-aware dense contact validation

Runs the **actual CPU C++ physics engine**, not a GPU kernel or animation. Kaggle Internet must be enabled to clone the repository. If Kaggle has a GPU attached, **this experiment still runs on CPU**. The benchmark is a 243-moving-box contact pile with 972 persistent contacts, **not one million interacting bodies**.

```python
import subprocess, os, pathlib, shutil
repo = pathlib.Path('/kaggle/working/avbd-demo3d')
if repo.exists(): shutil.rmtree(repo)
def run(args, **kwargs):
    print('+', ' '.join(map(str,args)),flush=True)
    return subprocess.run(args, check=True, **kwargs)
run(['git','clone','--depth','1','https://github.com/alxspiker/avbd-demo3d.git',str(repo)])
run(['cmake','-S',str(repo),'-B',str(repo/'build'),'-DCMAKE_BUILD_TYPE=Release','-DAVBD_ENABLE_OPENMP=ON'])
run(['cmake','--build',str(repo/'build'),'--parallel','4'])
```

```python
env=dict(os.environ,OMP_NUM_THREADS='4')
run(['ctest','--test-dir',str(repo/'build'),'--output-on-failure','-j','2'],env=env)
print('All tests passed')
```

```python
run([str(repo/'build/avbd3d_contact_allocations')],env=env)
run([str(repo/'build/avbd3d_contact_geometry')],env=env)
run([str(repo/'build/avbd3d_contact_benchmark'),'5','150'],env=env)
print('Stage 13 CPU dense-contact checks complete')
```
